```python
from dotenv import load_dotenv
import os
load_dotenv()
groq_key = os.getenv("GROQ_API_KEY")

import sys
sys.path.append('../src')
from tools import get_player_stats, predict_match, get_player_form_trend

from langchain_groq import ChatGroq
from langchain.agents import create_agent

llm = ChatGroq(model="openai/gpt-oss-120b", groq_api_key=groq_key, temperature=0.3)

system_prompt = """Tum ek cricket data assistant ho. Tumhare paas 3 tools hain: get_player_stats, predict_match, get_player_form_trend.

STRICT RULES:
1. Sirf tool se mila data use karo, apni taraf se koi stat, player naam, ya fact mat jodo.
2. Agar tool mein koi information nahi hai, toh use mat likho.
3. Kisi bhi player ya team ka naam mat likho jab tak wo tool ke output mein na ho.
4. Answer Hindi-English mix (Hinglish) mein do.
"""

tools_list = [get_player_stats, predict_match, get_player_form_trend]
agent = create_agent(llm, tools_list, system_prompt=system_prompt)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "Kohli ka form kaisa chal raha hai?"}]})
print(response["messages"][-1].content)
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()

agent = create_agent(llm, tools_list, system_prompt=system_prompt, checkpointer=checkpointer)
```

```python
config = {"configurable": {"thread_id": "test-conversation-1"}}

response1 = agent.invoke(
    {"messages": [{"role": "user", "content": "Kohli ka strike rate kitna hai?"}]},
    config=config
)
print(response1["messages"][-1].content)

print("\n---\n")

# Follow-up question, bina naam repeat kiye
response2 = agent.invoke(
    {"messages": [{"role": "user", "content": "aur uska form kaisa hai?"}]},
    config=config
)
print(response2["messages"][-1].content)
```
