```python
useful_cols = [
    'match_id', 'date', 'season', 'match_type', 'venue', 'city',
    'innings', 'batting_team', 'bowling_team',
    'over', 'ball', 'batter', 'bat_pos', 'bowler',
    'runs_batter', 'balls_faced', 'runs_total', 'runs_extras',
    'runs_not_boundary', 
    'extra_type',
    'wicket_kind', 'player_out',
    'toss_winner', 'toss_decision',
    'team_runs', 'team_balls', 'team_wicket', 'runs_target',
    'match_won_by', 'win_outcome', 'result_type',
    'player_of_match'
]

df = df[useful_cols]
df = df.drop_duplicates()

df.to_csv('../data/IPL_cleaned.csv', index=False)
print("Fixed! Final shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
df.head()
```

```python
df['season'] = df['season'].astype(str)
df['season'].unique()
```

```python
# removing the 2026 year or season of the ipl form the dataset

df = df[df['season'] != '2026']
df.shape
```

```python
df['season'].unique()
```

```python
season_map ={
    '2007/08' : 2008,
    '2009/10' : 2010,
    '2020/21' : 2020 
}

df['season'] = df['season'].astype(str)
df['season'] = df['season'].replace(season_map)
df['season'] = df['season'].astype(int)

sorted(df['season'].unique())
```

```python
df['city'] = df['city'].replace({"Bangalore" : "Bengaluru"})
(df['city'] == 'Unknown').sum()
df[df['city'] == 'Unknown']['venue'].unique()
```

```python
team_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Rising Pune Supergiant': 'Rising Pune Supergiants',
}

for col in ['batting_team', 'bowling_team', 'toss_winner', 'match_won_by', 'player_of_match']:
    if col in df.columns:
        df[col] = df[col].replace(team_mapping)

# Verify
print(sorted(df['batting_team'].unique()))
```
